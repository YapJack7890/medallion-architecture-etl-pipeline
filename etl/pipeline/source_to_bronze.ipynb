```python
from pathlib import Path
import pandas as pd
import shutil

# Bronze folder
bronze_folder = Path("../../Files/bronze")
bronze_folder.mkdir(parents=True, exist_ok=True)

# Archive folder
archive_folder = Path("../../Files/archive")
archive_folder.mkdir(parents=True, exist_ok=True)

for csv_file in folder.glob("*.csv"):

    file_name = csv_file.stem
    source_table = file_name.rsplit("_", 1)[0]
    table_name = source_table

    if table_name is None:
        print(f"Skipping {csv_file.name}")
        continue

    # Read CSV
    df = pd.read_csv(csv_file)

    # Bronze destination
    output_dir = bronze_folder / table_name
    output_dir.mkdir(parents=True, exist_ok=True)

    output_file = output_dir / f"{table_name}.parquet"

    # Save as Parquet
    df.to_parquet(output_file, index=False)

    print(f"{csv_file.name} -> {output_file}")

    # Move original CSV to archive
    archive_file = archive_folder / csv_file.name
    shutil.move(str(csv_file), str(archive_file))

    print(f"Archived: {archive_file}")
```

```python

def run_notebook(path):
    with open(path) as f:
        nb = nbformat.read(f, as_version=4)

    client = NotebookClient(nb)
    client.execute()

run_notebook("bronze_to_silver.ipynb")
```

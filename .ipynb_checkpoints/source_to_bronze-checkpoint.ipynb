```python
import pandas as pd 
from pathlib import Path
from sqlalchemy import create_engine
import os
```

```python

folder = Path(r"C:\Users\YapJack\Desktop\Data Science\University\Personal Projects\data-warehouse-project\source")

files = [file for file in folder.iterdir() if file.is_file()]

files
```

```python
TABLE_MAPPING = {
    "olist_customers_dataset.csv": "customers",
    "olist_geolocation_dataset.csv": "geolocation",
    "olist_orders_dataset.csv": "orders",
    "olist_order_payments_dataset.csv": "order_payments",
    "olist_order_items_dataset.csv": "order_items",
    "olist_products_dataset.csv": "products",
    "olist_order_reviews_dataset.csv": "order_reviews",
    "olist_sellers_dataset.csv": "sellers",
    "product_category_name_translation.csv": "product_category_translation",
}
```

```python
# PostgreSQL connection
engine = create_engine(
    "postgresql+psycopg2://postgres:password@localhost:5432/brazilian_ecommerce_dw"
)

for csv_file in folder.glob("*.csv"):
    table_name = TABLE_MAPPING.get(csv_file.name)

    if table_name is None:
        print(f"Skipping {csv_file.name}")
        continue

    df = pd.read_csv(csv_file)

    df.to_sql(
        table_name,
        engine,
        schema="bronze",
        if_exists="append",
        index=False
    )

    # print(f"{csv_file.name} -> bronze.{table_name}")
    
    if os.path.exists(csv_file):
        os.remove(csv_file)
        print("File inserted into Bronze Layer.")
    else:
        print("File does not exist.")
        
```

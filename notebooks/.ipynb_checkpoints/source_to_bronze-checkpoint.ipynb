```python
from pathlib import Path
import pandas as pd
import os

# Bronze folder
bronze_folder = Path("../Files/bronze")
bronze_folder.mkdir(exist_ok=True)

for csv_file in folder.glob("*.csv"):
    
    file_name = Path(csv_file).stem      # order_items_20260722
    source_table = file_name.rsplit("_", 1)[0]
    
    table_name = source_table

    if table_name is None:
        print(f"Skipping {csv_file.name}")
        continue

    # Read CSV
    df = pd.read_csv(csv_file)

    # Bronze destination
    output_dir = bronze_folder / table_name
    output_dir.mkdir(parents=True, exist_ok=True)

    output_file = output_dir / f"{table_name}.parquet"

    # Save as Parquet
    df.to_parquet(output_file, index=False)

    print(f"{csv_file.name} -> {output_file}")

    # Remove original CSV
    if os.path.exists(csv_file):
        os.remove(csv_file)
        print("File converted to Parquet and moved to Bronze Layer.")
    else:
        print("File does not exist.")

```

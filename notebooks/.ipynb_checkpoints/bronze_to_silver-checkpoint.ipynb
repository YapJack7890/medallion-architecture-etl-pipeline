```python
def remove_duplicate_rows(df: pd.DataFrame) -> pd.DataFrame:
    return df.drop_duplicates()


def remove_duplicate_keys(df: pd.DataFrame, primary_keys: list) -> pd.DataFrame:
    return df.drop_duplicates(subset=primary_keys)


def remove_null_primary_keys(df: pd.DataFrame, primary_keys: list) -> pd.DataFrame:
    return df.dropna(subset=primary_keys)


def replace_empty_with_null(df: pd.DataFrame) -> pd.DataFrame:
    return df.replace(r'^\s*$', pd.NA, regex=True)

    
def trim_string_values(df: pd.DataFrame) -> pd.DataFrame:
    object_cols = df.select_dtypes(include="object").columns

    for col in object_cols:
        df[col] = df[col].str.strip()

    return df
```

```python
import pandas as pd


def standardize_column_names(df: pd.DataFrame) -> pd.DataFrame:

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )

    return df


def convert_dates(df: pd.DataFrame, columns: list) -> pd.DataFrame:

    for col in columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")

    return df


def convert_numeric(df: pd.DataFrame, columns: list) -> pd.DataFrame:

    for col in columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    return df


def uppercase_columns(df: pd.DataFrame, columns: list) -> pd.DataFrame:

    for col in columns:
        df[col] = df[col].str.upper()

    return df


def rename_columns(df: pd.DataFrame, rename_map: dict) -> pd.DataFrame:

    return df.rename(columns=rename_map)
```

```python
import pandas as pd


def normalize_category_values(df, column, mapping):

    df[column] = df[column].replace(mapping)

    return df


def normalize_boolean(df, column):

    mapping = {
        "Y": True,
        "N": False,
        "Yes": True,
        "No": False,
        1: True,
        0: False
    }

    df[column] = df[column].replace(mapping)

    return df
```

```python
import pandas as pd


def add_order_year(df):

    df["order_year"] = df["order_purchase_timestamp"].dt.year

    return df


def add_order_month(df):

    df["order_month"] = df["order_purchase_timestamp"].dt.month

    return df


def add_order_day(df):

    df["order_day"] = df["order_purchase_timestamp"].dt.day

    return df


def calculate_delivery_days(df):

    df["delivery_days"] = (
        df["order_delivered_customer_date"] -
        df["order_purchase_timestamp"]
    ).dt.days

    return df
```

```python
import pandas as pd


def add_product_category_english(df, translation_df):

    return df.merge(
        translation_df,
        on="product_category_name",
        how="left"
    )


def add_customer_location(df, geolocation_df):

    return df.merge(
        geolocation_df,
        on="geolocation_zip_code_prefix",
        how="left"
    )
```
